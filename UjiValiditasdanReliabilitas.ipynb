```python
import re
import pandas as pd
import numpy as np
from scipy import stats
from google.colab import files
```

```python
up = files.upload()

nama_file = list(up.keys())[0]

print("File terbaca:", nama_file)
```

```python
df = pd.read_csv(nama_file)

# Mengambil kolom yang merupakan pernyataan kuesioner
kolom_butir = [
    kolom for kolom in df.columns
    if "[" in kolom and "]" in kolom
]

# Memisahkan data pernyataan
butir = df[kolom_butir]

print("Jumlah responden :", len(df))
print("Jumlah seluruh kolom :", len(df.columns))
print("Jumlah butir :", butir.shape[1])
print("Data kosong :", int(butir.isna().sum().sum()))

display(butir.head())
```

```python
skala = {
    "Sangat Tidak Setuju (STS)": 1,
    "Tidak Setuju (TS)": 2,
    "Netral (N)": 3,
    "Setuju (S)": 4,
    "Sangat Setuju (SS)": 5
}

num = butir.replace(skala).apply(
    pd.to_numeric,
    errors="coerce"
)

print(
    "Sel yang gagal dikonversi:",
    int(num.isna().sum().sum())
)

display(num.head())
```

```python
dimensi = {}

for kolom in butir.columns:

    nama = re.match(
        r"^(.*?)\s*\[",
        kolom
    ).group(1).strip().rstrip(".").strip()

    # Menyamakan penulisan nama dimensi
    if nama == "Rsspon emosional":
        nama = "Respon emosional"

    dimensi.setdefault(nama, []).append(kolom)


print("HASIL PENGELOMPOKAN DIMENSI\n")

for nama, daftar in dimensi.items():

    print(f"{nama}: {len(daftar)} butir")

print("\nJumlah dimensi:", len(dimensi))
print("Jumlah seluruh butir:", sum(len(v) for v in dimensi.values()))
```

```python
dimensi = {}

for kolom in butir.columns:

    hasil = re.match(r"^(.*?)\s*\[", kolom)

    if hasil:
        nama = hasil.group(1).strip()

        dimensi.setdefault(nama, []).append(kolom)


# Pembagian Topic
pembagian_topic = {
    "Penggunaan AI pada tugas atau projek kuliah": "Topic 1",

    "Aspek Psikologis penggunaan AI pada tugas atau projek kuliah": "Topic 1",

    "Kebiasaan penggunaan teknologi digital": "Topic 2",

    "Prokrastinasi akademik.": "Topic 2",

    "Rsspon emosional": "Topic 2"
}


for nama, daftar in dimensi.items():

    topic = pembagian_topic.get(nama, "Belum ditentukan")

    print(f"\n{topic} - {nama}: {len(daftar)} butir")
```

```python
n = len(num)

t = stats.t.ppf(0.975, n - 2)

r_tabel = t / np.sqrt((n - 2) + t**2)

print(
    f"n = {n} | "
    f"r tabel (α=0,05; df={n-2}) = {r_tabel:.3f}"
)


hasil_validitas = []

for nama, daftar in dimensi.items():

    x = num[daftar]

    total = x.sum(axis=1)

    for i, kolom in enumerate(daftar, 1):

        r, p = stats.pearsonr(x[kolom], total)

        r_koreksi, _ = stats.pearsonr(
            x[kolom],
            total - x[kolom]
        )

        hasil_validitas.append({

            "Topic": pembagian_topic[nama],

            "Dimensi": nama,

            "Butir": f"B{i}",

            "r hitung": round(r, 3),

            "Corrected r": round(r_koreksi, 3),

            "p-value": round(p, 4),

            "Keterangan": (
                "Valid" if r > r_tabel
                else "Tidak valid"
            )

        })


tabel_validitas = pd.DataFrame(hasil_validitas)

display(tabel_validitas)
```

```python
def cronbach_alpha(x):

    k = x.shape[1]

    if k < 2:
        return np.nan

    varians_butir = x.var(ddof=1).sum()

    varians_total = x.sum(axis=1).var(ddof=1)

    if varians_total == 0:
        return np.nan

    return (
        k / (k - 1)
        * (1 - varians_butir / varians_total)
    )


def kategori(a):

    if pd.isna(a):
        return "Tidak dapat dihitung"

    return (
        "Sangat baik" if a >= 0.90
        else "Baik" if a >= 0.80
        else "Reliabel" if a >= 0.70
        else "Tidak reliabel"
    )


hasil_reliabilitas = []

alpha_dihapus = []


for nama, daftar in dimensi.items():

    x = num[daftar]

    a = cronbach_alpha(x)

    hasil_reliabilitas.append({

        "Topic": pembagian_topic[nama],

        "Dimensi": nama,

        "Jumlah butir": len(daftar),

        "Cronbach's alpha": round(a, 3),

        "Kategori": kategori(a)

    })


    # Menghitung alpha jika butir dihapus

    for i, kolom in enumerate(daftar, 1):

        if len(daftar) > 2:

            alpha_baru = cronbach_alpha(
                x.drop(columns=kolom)
            )

            alpha_dihapus.append({

                "Topic": pembagian_topic[nama],

                "Dimensi": nama,

                "Butir": f"B{i}",

                "Alpha jika dihapus": round(alpha_baru, 3)

            })


tabel_reliabilitas = pd.DataFrame(
    hasil_reliabilitas
)

tabel_alpha_dihapus = pd.DataFrame(
    alpha_dihapus
)


print("HASIL UJI RELIABILITAS")

display(tabel_reliabilitas)


print("ALPHA JIKA BUTIR DIHAPUS")

display(tabel_alpha_dihapus)
```

```python
gabung = tabel_validitas.merge(
    tabel_alpha_dihapus[
        ["Dimensi", "Butir", "Alpha jika dihapus"]
    ],
    on=["Dimensi", "Butir"],
    how="left"
)


with pd.ExcelWriter(
    "hasil_uji_validitas_reliabilitas.xlsx"
) as w:

    gabung.to_excel(
        w,
        sheet_name="Validitas",
        index=False
    )

    tabel_reliabilitas.to_excel(
        w,
        sheet_name="Reliabilitas",
        index=False
    )


files.download(
    "hasil_uji_validitas_reliabilitas.xlsx"
)
```

```python
# Menampilkan butir yang tidak valid

butir_tidak_valid = tabel_validitas[
    tabel_validitas["Keterangan"] == "Tidak valid"
]

print("BUTIR YANG TIDAK VALID")
display(butir_tidak_valid[
    ["Dimensi", "Butir", "r hitung", "Corrected r", "p-value"]
])

print("Nilai r tabel:", round(r_tabel, 3))

print(
    "Jumlah butir tidak valid:",
    len(butir_tidak_valid)
)
```

```python
# Menghitung jumlah butir valid berdasarkan r hitung
valid_r_hitung = (
    tabel_validitas["r hitung"] > r_tabel
).sum()

# Menghitung jumlah butir valid berdasarkan corrected r
valid_corrected = (
    tabel_validitas["Corrected r"] >= 0.30
).sum()

# Membuat tabel perbandingan
perbandingan = pd.DataFrame({
    "Kriteria": [
        "r hitung > r tabel",
        "Corrected r >= 0,30"
    ],
    "Jumlah butir valid": [
        valid_r_hitung,
        valid_corrected
    ]
})

display(perbandingan)

print(
    "Selisih jumlah butir:",
    abs(valid_r_hitung - valid_corrected)
)
```

```python
# Memilih satu butir tidak valid
if len(butir_tidak_valid) > 0:

    target = butir_tidak_valid.iloc[0]

    nama_dimensi = target["Dimensi"]
    nomor_butir = target["Butir"]

    daftar_butir = dimensi[nama_dimensi]

    kolom_dihapus = daftar_butir[
        int(nomor_butir.replace("B", "")) - 1
    ]

    # Data sebelum penghapusan
    data_sebelum = num[daftar_butir]

    alpha_sebelum = cronbach_alpha(data_sebelum)

    # Data sesudah penghapusan
    daftar_baru = [
        k for k in daftar_butir
        if k != kolom_dihapus
    ]

    data_sesudah = num[daftar_baru]

    alpha_sesudah = cronbach_alpha(data_sesudah)

    hasil_perbaikan = pd.DataFrame({
        "Keterangan": [
            "Sebelum penghapusan",
            "Sesudah penghapusan"
        ],
        "Jumlah butir": [
            len(daftar_butir),
            len(daftar_baru)
        ],
        "Cronbach's Alpha": [
            round(alpha_sebelum, 3),
            round(alpha_sesudah, 3)
        ]
    })

    print("Dimensi:", nama_dimensi)
    print("Butir yang dihapus:", nomor_butir)

    display(hasil_perbaikan)

else:
    print("Tidak ditemukan butir tidak valid.")
```

```python
# Mengambil sampel acak sebanyak 20 responden
num_sample = num.sample(
    n=20,
    random_state=1
)

# Menghitung r tabel sampel
n_sample = len(num_sample)

t_sample = stats.t.ppf(
    0.975,
    n_sample - 2
)

r_tabel_sample = t_sample / np.sqrt(
    (n_sample - 2) + t_sample**2
)

# Menguji validitas ulang
hasil_sample = []

for nama, daftar in dimensi.items():

    x = num_sample[daftar]
    total = x.sum(axis=1)

    for i, kolom in enumerate(daftar, 1):

        r, p = stats.pearsonr(
            x[kolom],
            total
        )

        hasil_sample.append({
            "Dimensi": nama,
            "Butir": f"B{i}",
            "r hitung": round(r, 3),
            "Keterangan": (
                "Valid"
                if r > r_tabel_sample
                else "Tidak valid"
            )
        })

tabel_sample = pd.DataFrame(hasil_sample)

print("Jumlah sampel:", n_sample)
print("r tabel sampel:", round(r_tabel_sample, 3))

display(tabel_sample)

print(
    "Jumlah butir valid:",
    (tabel_sample["Keterangan"] == "Valid").sum()
)

print(
    "Jumlah butir tidak valid:",
    (tabel_sample["Keterangan"] == "Tidak valid").sum()
)
```
